### 1. Data Resampling & Aggregation
To reduce noise and make the dataset computationally manageable, we resample the minute-by-minute data to a daily frequency (OHLCV).
This reduces our dataset size from 7.6M rows to approximately 5,290 rows without losing the overall long-term trend.


```python
# Set index to Datetime
df.set_index('Datetime', inplace=True)

# Resample to daily OHLCV
df_daily = df.resample('D').agg({
    'Open': 'first',
    'High': 'max',
    'Low': 'min',
    'Close': 'last',
    'Volume': 'sum'
})

print('Resampled Daily Dataset Shape:', df_daily.shape)
print(df_daily.head())
```

### 2. Feature Engineering
We create a set of predictive features including price/volume lags, rolling moving averages, daily spread, and rolling volatility.


```python
# Create Lags
for lag in [1, 2, 3, 7, 14, 30]:
    df_daily[f'Close_Lag_{lag}'] = df_daily['Close'].shift(lag)
    df_daily[f'Volume_Lag_{lag}'] = df_daily['Volume'].shift(lag)

# Create Moving Averages
for window in [7, 30, 90]:
    df_daily[f'Close_MA_{window}'] = df_daily['Close'].rolling(window=window).mean()
    df_daily[f'Volume_MA_{window}'] = df_daily['Volume'].rolling(window=window).mean()

# High-Low spread
df_daily['HL_Spread'] = df_daily['High'] - df_daily['Low']
df_daily['HL_Spread_Lag_1'] = df_daily['HL_Spread'].shift(1)

# Returns & Volatility
df_daily['Returns'] = df_daily['Close'].pct_change()
df_daily['Volatility_7d'] = df_daily['Returns'].rolling(7).std()
df_daily['Volatility_30d'] = df_daily['Returns'].rolling(30).std()

# Drop NaNs created by rolling windows/lags
df_daily.dropna(inplace=True)
print('Cleaned dataset shape:', df_daily.shape)
```

### 3. Chronological Train-Test Split
We perform a time-based train-test split to avoid look-ahead bias and replicate real-world performance. 
- **Train Set**: Data from 2012 to 2024-12-31
- **Test Set**: Data from 2025-01-01 onwards (~1.5 years of out-of-sample data)


```python
# Split criteria
train_df = df_daily.loc[df_daily.index < '2025-01-01']
test_df = df_daily.loc[df_daily.index >= '2025-01-01']

print(f'Train set: {train_df.index.min()} to {train_df.index.max()} ({len(train_df)} days)')
print(f'Test set: {test_df.index.min()} to {test_df.index.max()} ({len(test_df)} days)')
```

### 4. Experiment 1: Forecasting Raw Close Price
In our first experiment, we define our target variable directly as next day's raw Close price ($Close_{t+1}$).
We compare a Naive Baseline, Linear Regression, and a Random Forest Regressor.


```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Define raw target (shift Close back by 1 day)
df_daily['Target_Raw'] = df_daily['Close'].shift(-1)
df_raw = df_daily.dropna(subset=['Target_Raw']).copy()

# Define features list (exclude targets and OHLCV values to prevent data leakage)
features = [col for col in df_raw.columns if col not in ['Target_Raw', 'Target_Diff', 'Target_Price', 'Open', 'High', 'Low', 'Close', 'Volume', 'HL_Spread', 'Returns', 'Price_Diff']]

X = df_raw[features]
y = df_raw['Target_Raw']

X_train_raw = X.loc[X.index < '2025-01-01']
y_train_raw = y.loc[y.index < '2025-01-01']
X_test_raw = X.loc[X.index >= '2025-01-01']
y_test_raw = y.loc[y.index >= '2025-01-01']

# Naive Baseline prediction: Tomorrow's price = Today's Close price
naive_pred = df_raw.loc[df_raw.index >= '2025-01-01', 'Close']

results_raw = {
    'Naive Baseline': {
        'MAE': mean_absolute_error(y_test_raw, naive_pred),
        'RMSE': np.sqrt(mean_squared_error(y_test_raw, naive_pred)),
        'R2': r2_score(y_test_raw, naive_pred)
    }
}

models_raw = {
    'Linear Regression': LinearRegression(),
    'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
}

for name, model in models_raw.items():
    model.fit(X_train_raw, y_train_raw)
    preds = model.predict(X_test_raw)
    results_raw[name] = {
        'MAE': mean_absolute_error(y_test_raw, preds),
        'RMSE': np.sqrt(mean_squared_error(y_test_raw, preds)),
        'R2': r2_score(y_test_raw, preds)
    }

res_raw_df = pd.DataFrame(results_raw).T
print('Raw Price Prediction Results:')
print(res_raw_df.to_string())
```

#### Why did the Random Forest Regressor fail on the test set?
Tree-based regressor algorithms make predictions by averaging the target values in their leaf nodes. Consequently, they **cannot extrapolate trends** or predict values higher or lower than the range observed during training. 
Since Bitcoin reached new all-time highs in 2025 and 2026, the Random Forest model's predictions were capped at the peak price from 2012-2024, leading to poor R2 performance compared to the Naive Baseline and Linear Regression.


### 5. Experiment 2: Forecasting Price Differences
To resolve the extrapolation issue, we train our models on the daily **price difference** ($Close_{t+1} - Close_t$). 
Since price difference is a stationary feature, Random Forest can learn the delta predictions effectively, and we can then reconstruct the absolute Close price by adding the predicted delta to today's price.


```python
# Daily Price Difference
df_daily['Price_Diff'] = df_daily['Close'].diff()

# Target: Next day's price change
df_daily['Target_Diff'] = df_daily['Price_Diff'].shift(-1)
df_daily['Target_Price'] = df_daily['Close'].shift(-1)

df_diff = df_daily.dropna(subset=['Target_Diff']).copy()

X_diff = df_diff[features]
y_train_diff = df_diff.loc[df_diff.index < '2025-01-01', 'Target_Diff']
X_test_diff = df_diff.loc[df_diff.index >= '2025-01-01', features]
y_test_price = df_diff.loc[df_diff.index >= '2025-01-01', 'Target_Price']
y_train_price = df_diff.loc[df_diff.index < '2025-01-01', 'Target_Price']
X_train_diff = df_diff.loc[df_diff.index < '2025-01-01', features]

results_diff = {
    'Naive Baseline': results_raw['Naive Baseline']
}

models_diff = {
    'Linear Regression (Diff)': LinearRegression(),
    'Random Forest (Diff)': RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
}

for name, model in models_diff.items():
    model.fit(X_train_diff, y_train_diff)
    pred_diffs = model.predict(X_test_diff)
    
    # Reconstruct absolute price predictions: Today's Close + Predicted Diff
    pred_prices = df_diff.loc[df_diff.index >= '2025-01-01', 'Close'] + pred_diffs
    
    results_diff[name] = {
        'MAE': mean_absolute_error(y_test_price, pred_prices),
        'RMSE': np.sqrt(mean_squared_error(y_test_price, pred_prices)),
        'R2': r2_score(y_test_price, pred_prices)
    }

res_diff_df = pd.DataFrame(results_diff).T
print('Price Difference Prediction Results:')
print(res_diff_df.to_string())
```

### 6. Visualizing Model Forecast Comparison
We plot the actual closing price against the forecasted closing price for all models on our test set.


```python
plt.figure(figsize=(12, 6))

# Actual test set prices
test_dates = y_test_price.index
plt.plot(test_dates, y_test_price, label='Actual Price', color='black', linewidth=1.5)

# Naive Baseline
plt.plot(test_dates, naive_pred, label='Naive Baseline', color='gray', linestyle='--', alpha=0.8)

# Linear Regression (Diff)
lr_diff_model = models_diff['Linear Regression (Diff)']
lr_preds = df_diff.loc[df_diff.index >= '2025-01-01', 'Close'] + lr_diff_model.predict(X_test_diff)
plt.plot(test_dates, lr_preds, label='Linear Regression (Diff)', color='orange', alpha=0.8)

# Random Forest (Diff)
rf_diff_model = models_diff['Random Forest (Diff)']
rf_preds = df_diff.loc[df_diff.index >= '2025-01-01', 'Close'] + rf_diff_model.predict(X_test_diff)
plt.plot(test_dates, rf_preds, label='Random Forest (Diff)', color='green', alpha=0.8)

plt.title('Bitcoin Price Forecast Comparison on Test Set (2025 - June 2026)')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.3)
plt.show()
```

### 7. Feature Importance Analysis
Finally, we extract the feature importances from our trained Random Forest model to see which indicators have the most predictive power.


```python
importances = rf_diff_model.feature_importances_
indices = np.argsort(importances)

plt.figure(figsize=(10, 6))
plt.title('Random Forest Feature Importances (Price Diff Model)')
plt.barh(range(len(importances)), importances[indices], align='center', color='teal')
plt.yticks(range(len(importances)), [features[i] for i in indices])
plt.xlabel('Relative Importance')
plt.grid(True, axis='x', linestyle=':', alpha=0.3)
plt.show()
```
